# *args_schema

```python
from langchain_core.utils.function_calling import convert_to_openai_tool
from langchain_core.tools import tool
from rich import print as rprint
from pydantic import BaseModel , Field
from typing import Literal

class WeatherInput(BaseModel):
    city:str = Field(
        description="城市名称，如北京",
        default="bzjy"
    )
    unit: Literal["Celsius","Fahrenheit"] = Field(
        default="Celsius",
    )

@tool(args_schema=WeatherInput)
def get_weather(city:str ,     unit: Literal["Celsius","Fahrenheit"]="Celsius"):
    """
    获取指定城市的天气信息
    """

    return f"{city} ,{unit}+ 15度"


rprint(convert_to_openai_tool(get_weather))
```

# *json Schema

```python
json_schema =  {
            'properties': {
                'city': {'default': 'bzjy', 'description': '城市名1称，如北京', 'type': 'string'},
                'unit': {'default': 'Celsius', 'enum': ['Celsius', 'Fahrenheit'], 'type': 'string'}
            },
            'type': 'object'
        }

@tool(args_schema=json_schema)
def get_weather(city:str):
    """
    获取指定城市的天气信息
    """

    return f"{city}+ 15度"

rprint(convert_to_openai_tool(get_weather))
```
