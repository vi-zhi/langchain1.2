```python
import os

from dotenv import load_dotenv
from langchain_deepseek import ChatDeepSeek
from openai.resources.skills import content

load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")
llm_deepseek = ChatDeepSeek(
    model="deepseek-v4-flash",
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
)
```

```python
def keep_recent_messages(message , max_pairs = 3):
    system_message = [m for m in message if m.get("role") == "system"]
    conversation_message = [m for m in message if m.get("role") != "system"]

    recent_message = conversation_message[-(max_pairs * 2):]

    return system_message + recent_message
```

```python
long_conversation = [
    {"role":"system" , "content":"你是python导师"}
]
# 第 1 轮
long_conversation.append({"role": "user", "content": "什么是列表？用一句解释"})
r1 = llm_deepseek.invoke(long_conversation)
long_conversation.append({"role": "assistant", "content": r1.content})

# 第 2 轮
long_conversation.append({"role": "user", "content": "列表和元组有什么区别？用一句解释"})
r2 = llm_deepseek.invoke(long_conversation)
long_conversation.append({"role": "assistant", "content": r2.content})

# 第 3 轮
long_conversation.append({"role": "user", "content": "什么是字典呢？用一句解释"})
r3 = llm_deepseek.invoke(long_conversation)
long_conversation.append({"role": "assistant", "content": r3.content})

print(len(long_conversation))

optimized = keep_recent_messages(long_conversation,max_pairs = 2)

print(len(optimized))

# 第 4 轮
optimized.append({"role": "user", "content": "我问的第一个问题是啥"})
r4 = llm_deepseek.invoke(optimized)
long_conversation.append({"role": "assistant", "content": r4.content})

print(r4.content)
```

```python
import os

from dotenv import load_dotenv
from langchain_deepseek import ChatDeepSeek

load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")
llm_deepseek = ChatDeepSeek(
    model="deepseek-v4-flash",
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
)

messages=[
    {"role":"system" , "content":"你是一个ai助手，帮助用户解决各种问题"}
]
i = 1
while True:
    print("\n","="*10,f"第{i}轮对话","="*10)

    user_message = input("请输入：")

    if user_message == "quit":
        print("="*10,"exit","="*10)
        break

    messages.append({"role":"user" , "content":user_message})

    print("ai:",end="",flush=True)

    memory_message=keep_recent_messages(messages , 10)

    tem_message=[]

    for ai_message in llm_deepseek.stream(memory_message):
        print(ai_message.content,end="",flush=True)
        tem_message += ai_message
    print("\n","="*10,f"第{i}轮对话end","="*10)
    i += 1
    memory_message.append({"role":"assistant" , "content":tem_message})
```
