## 1.deepseek

```python
import os
from tempfile import template

from dotenv import load_dotenv
from langchain_deepseek import ChatDeepSeek
from langchain_openai import ChatOpenAI

load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")
llm_deepseek = ChatDeepSeek(
    model="deepseek-v4-flash",
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
)

res = llm_deepseek.invoke("你好")
print(res)
```

## 2.openai
  调用deepseek大模型

```python
load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

model = ChatOpenAI(
    model="deepseek-v4-flash",
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
)

res = model.invoke("1+1=?")
print(res.content)
```

## 3.使用langchain1.x

```python
import os

from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")
llm_deepseek = init_chat_model(
    model="deepseek-v4-flash",
    model_provider="deepseek",
    temperature=0.3,
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
)

response = model.invoke("张三，男，30岁，拥有8年编程开发经验，目前在某互联网大厂担任技术专家。帮我从上文中提取数据，返回JSON格式")
print(response.content)
```

```python
model.invoke("翻译你好世界")
```

```python
message = [
    {"role":"system" , "content":"你是一个专业的数学老师"},
    {"role":"user" , "content":"我叫小明"}
]

res = model.invoke(message)
print(res)
```

```python
conversation = [
    {"role":"system" , "content":"你是一个专业的数学老师"},
    {"role":"user" , "content":"我叫小明"}
]

res = model.invoke(message)
print(res)
conversation.append({"role":"assistant" , "content":res.content})
conversation.append({"role":"user" , "content":"我叫什么"})
result = model.invoke(conversation)
print("--------------------------------------------------")
print(result)

```

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
from rich import print as rprint
conversation = [
    SystemMessage(content="你是一个专业的数学老师"),
    HumanMessage(content="解释斐波那契数列"),
]

res = model.invoke(message)
print(res.content)
conversation.append(AIMessage(content=res.content))
conversation.append(HumanMessage(content="我刚才的问题是啥"))
result = model.invoke(conversation)
print("--------------------------------------------------")
print(result)

rprint(result)
```
